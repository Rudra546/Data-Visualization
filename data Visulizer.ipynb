```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class Visulizer:
    def __init__(self):
        self.data = None
```

```python
def Load_data(self):
    path = input("Enter Your Csv File Path: ")
    try:
        if path.endswith(".csv"):
            self.data = pd.read_csv(path)
            print("Data Loaded Successfully...!")
            print(self.data.head())
        else:
            print("Unsupported File Conversion")
    except FileNotFoundError:
        print("File ot Found Please Chaeck Your Path")
    except Exception as e:
        print(f"failed to load Data Set")

Visulizer.Load_data=Load_data
```

```python
def Explore_Data(self):
    print("1.Display The First 5 Rows")
    print("2.Display The Last 5 Columns")
    print("3.Display Columns Name")
    print("4.Display Data Types")
    print("5.Display Basic info\n")

    try:
        choice = int(input("Enter Your Choise"))

        if choice == 1:
            print(self.data.head())
        elif choice == 2:
            print(self.data.tail())
        elif choice == 3:
            print(self.data.columns)
        elif choice == 4:
            print(self.data.dtypes)
        elif choice == 5:
            print(self.data.info())
        else:
            print("Invaild Choice please Try Again and enter between 1 or 5.")
    except Exception as e:
        print(f"Failed to Explore DataSet : {e}")

Visulizer.Explore_Data=Explore_Data
```

```python
def df_operations(self):
    print("Perform Data Frame Operations")
    print("1.Filter By Integer Data")
    print("2.Filtered By Float Data")
    print("3.Groupby")
    print("4.Min Value")
    print("5.Max Value")

    try:
        choice = int(input("Enter Your Choice: "))

        if choice == 1:
            col = input("Enter Iteger Column Name :")
            val = int(input("Enter The Value:"))
            print(self.data[self.data[col] > Val])
        elif choice == 2:
            column = input("Enter your float column name to filter: ")
            value = float(input("Enter value to filter: "))
            print(self.data[self.data[column] > value])
        elif choice == 3:
            column = input("Enter the code Groupby")
            if column in self.data.columns:
                print(self.data.groupby(column).size())
            else:
                print("Column Not Found")
        elif choice == 4:
            print("Minimum Value:\n",self.data.min(numeric_only=True))
        elif choice == 5:
            print("Maximum Value:\n",self.data.max(numeric_only=True))
        else:
            print("Invalid Choice Enter 1 to 5.")
    except Exception as e:
        print(f"Error:{e}")

Visulizer.df_operations=df_operations
```

```python
def Missing_value(self):
    print("Handle Missing Value Data")
    print("1. display missing values count")
    print("2. fill missing values with mean")
    print("3. drop rows with missing values")
    print("4. replace missing values with specific value\n")

    try:
        choice = int(input("Enter the choice : "))

        if choice == 1:
            print(self.data.isnull().sum())
        elif choice == 2:
            print("It only fills then numerical columns")
            label = input("Enter Column Name:")
            self.data[label].fillna(self.data[label].mean(),inplace=True)
            print(self.data[label])
        elif choice == 3:
            label = input("Enter The drop With Missing Value:")
            self.data = self.data[self.data[label].notna()]
            print("Null Value Removed..!")
        elif choice == 4:
            value = input("Enter the value into replaced missing value with specific value:")
            self.data.fillna(value,inplace=True)
            print(self.data)
        else:
            print("Invalid choice...!")
    except Exception as e:
        print(f"Failed To Handle Missing Value...!")

Visulizer.Missing_value=Missing_value
```

```python
def genrate(self):
    print(self.data.describe())

Visulizer.genrate=genrate
```

```python
def Data_visulization(self):
    print("Data Visulization")
    print("1.Bar Plot")
    print("2.Line Plot")
    print("3.Scatter Plot")
    print("4.Pie Chart")
    print("5.Histrogram")
    print("6.Stack Plot")

    try:
        choice = int(input("Enter the choice: "))

        if choice == 1:
            A = input("Enter X-Axis Columns:") 
            B = input("Enter Y-Axis Columns:")
            C = input("Enter title")
            D = input("Enter Legend Name:")
            sns.barplot(data=self.data,x=A,y=B,label=D)
            plt.title(C)
            plt.legend()
            plt.show()
        elif choice == 2:
            E = input("Enter X-Axis Columns:") 
            F = input("Enter Y-Axis Columns:")
            G = input("Enter title")
            H = input("Enter Legend Name:")
            plt.plot(self.data[E],self.data[F],linestyle = "--",marker = "o")
            plt.title(G)
            plt.legend([H])
            plt.show()
        elif choice == 3:
            I = input("Enter X-Axis Columns:") 
            J = input("Enter Y-Axis Columns:")
            K = input("Enter Hiu Columns:")
            sns.scatterplot(data=self.data,x=I,y=J,hue=K)
            plt.show()
        elif choice == 4:
            L = input("Enter Column Name:")
            plt.pie(self.data[A],autopct="%.2f%%")
            plt.show()
        elif choice == 5:
            M = input("Enter Col Name:")
            sns.histplot(self.data[M])
            plt.show()
        elif choice == 6:
            N = input("Enter X-Axis Columns:")
            O = input("Enter Y-Axis Columns:")
            plt.stackplot(self.data[N],self.data[O])
            plt.show()
        else:
            print("Invalid choice...!")
    except Exception as e:
        print(f"Failed To visualized data:{e}")

Visulizer.Data_visulization=Data_visulization
```

```python
def Save_vsiulization(self):
    A = input("Enter X-Axis Columns:") 
    B = input("Enter Y-Axis Columns:")
    C = input("Enter title")
    FileName = input("Enter FileName(e.g.,plot.png):")
    plt.plot(self.data[A],self.data[B])
    plt.title(C)
    plt.savefig(FileName,dpi=4000,bbox_inches = 'tight')
    print(f"Plot saved As {FileName}")

Visulizer.Save_vsiulization=Save_vsiulization
```

```python
X = Visulizer()
```
